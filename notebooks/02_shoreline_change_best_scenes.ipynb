# 02. Shoreline extraction, co-registration and change rates (best scene per year)

Runs `src/kerachut_shoreline.py` on the GeoTIFFs exported by notebook 01 (one clearest scene per year).

Inputs: a Google Drive folder holding `Kerachut_YYYYMMDD_SENSOR.tif` files (5 bands: blue, green, red, nir, swir1; UTM 47N).
Outputs (in `results/`): co-registration table, shoreline position per transect, change rates, GeoJSON shorelines and transects, figures 1 to 3.

Note: Earth Engine sometimes writes simultaneous exports into several Drive folders with the same name. Move all the .tif files into one folder before running.
